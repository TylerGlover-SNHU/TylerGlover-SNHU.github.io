### 2 Hidden Layers (Text Example)

```python
from __future__ import print_function
import numpy as np
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers.core import Dense, Activation
from keras.optimizers import SGD
from keras.utils import np_utils
np.random.seed(1671) # for reproducibility
# network and training
NB_EPOCH = 20
BATCH_SIZE = 128
VERBOSE = 1
NB_CLASSES = 10 # number of outputs = number of digits
OPTIMIZER = SGD()
N_HIDDEN = 128
VALIDATION_SPLIT=0.2 # how much TRAIN is reseved for VALIDATION
# data: shuffled and split between train and tests sets
(X_train, y_train), (X_test, y_test) = mnist.load_data()
#X_train is 60000 rows of 28x28 values --> reshaped in 60000 x 784
RESHAPED = 784
X_train = X_train.reshape(60000, RESHAPED)
X_test = X_test.reshape(10000, RESHAPED)
X_train = X_train.astype('float32')
X_test = X_test.astype('float32')
# normalize
X_train /= 255
X_test /= 255
print(X_train.shape[0], 'train samples')
print(X_test.shape[0], 'test samples')
# convert class vectors to binary class matrices
Y_train = np_utils.to_categorical(y_train, NB_CLASSES)
Y_test = np_utils.to_categorical(y_test, NB_CLASSES)
# M_HIDDEN hidden layers
# 10 outputs
# final state is softmax
model = Sequential()
# first layer
model.add(Dense(N_HIDDEN, input_shape=(RESHAPED,)))
model.add(Activation('relu'))
# second layer
model.add(Dense(N_HIDDEN))
model.add(Activation('relu'))
model.add(Dense(NB_CLASSES))
model.add(Activation('softmax'))
model.summary()
model.compile(loss='categorical_crossentropy', optimizer=OPTIMIZER, 
              metrics=['accuracy'])
history = model.fit(X_train, Y_train, batch_size=BATCH_SIZE, epochs=NB_EPOCH,
                    verbose=VERBOSE, validation_split=VALIDATION_SPLIT)
score = model.evaluate(X_test, Y_test, verbose=VERBOSE)
print("Test score:", score[0])
print('Test accuracy:', score[1])
```

## Evaluating Different Numbers of Layers

Adding additional layers to the model adds additional complexity since the output of one layer is passed to the next for further processing. In the case of the added layers for this assignment, the layers' structure are similar. As I learn more about how to construct layers in a model, I would expect to be able to add additional functionality. We would expect a higher level of accuracy as more layers are added. This is because deeper layers allow the model to further abstract the information processed and allows the model to handle more nuance. The training, validation, and test accurcies are given in the table below.

| Number of Layers | Training Accuracy | Validation Accuracy | Test Accuracy |
| --- | --- | --- | --- |
| 1 | 92.96% | 93.17% | 93.40% |
| 2 | 94.80% | 94.87% | 94.79% |
| 3 | 95.50% | 95.68% | 95.26% |
| 4 | 96.13% | 95.97% | 95.72% |

As the number of layers increases, we see that all accuracies increase. The rate of increase suggessts diminishing returns for the added complexity. We can see that training accuracy appears to grow more quickly than Validation and Test accuracy. This could be a symptom of overfitting the model. Since the model is trained on the test data, a model that is overfit would continue to get better at predicting training examples as it captures the specifics of the training data. However, if the model is overfit, it may get more tests wrong for the data not from the training set.
